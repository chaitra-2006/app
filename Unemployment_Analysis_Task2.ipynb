```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

print("Task 2 Started")

dates = pd.date_range('2019-01-01', '2020-11-30', freq='M')
np.random.seed(42)
rates = []
for d in dates:
    if d.year == 2020 and d.month in [4,5,6]:
        rates.append(np.random.uniform(20, 24))
    elif d.year == 2020 and d.month in [3,7]:
        rates.append(np.random.uniform(10, 15))
    else:
        rates.append(np.random.uniform(4, 9))

df = pd.DataFrame({'Date': dates, 'Unemployment_Rate': rates})
df['Year'] = df['Date'].dt.year

print(df.head())
print("Shape:", df.shape)

plt.figure(figsize=(12,5))
plt.plot(df['Date'], df['Unemployment_Rate'], marker='o', color='red', linewidth=2)
plt.title("Unemployment Rate Over Time - COVID Impact")
plt.ylabel("Unemployment Rate (%)")
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

plt.figure(figsize=(8,5))
yearly = df.groupby('Year')['Unemployment_Rate'].mean()
yearly.plot(kind='bar', color=['skyblue', 'tomato'])
plt.title("Yearly Avg - 2020 Spike Due to COVID")
plt.ylabel("Avg Unemployment %")
plt.show()

print("TASK 2 COMPLETE!")
```
